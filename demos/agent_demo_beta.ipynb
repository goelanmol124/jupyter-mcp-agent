# Companion demo: Beta-Binomial posterior

**Beta** prior plus **Binomial** likelihood: conjugate analytic posterior vs MLE \(k/n\), with a Monte Carlo sanity check.


```python
import numpy as np
rng = np.random.default_rng(11)
```

```python
prior_a, prior_b = 2.0, 2.0
n_trials = 40
k_success = 12
post_a = prior_a + k_success
post_b = prior_b + (n_trials - k_success)
```

```python
mean_p = post_a / (post_a + post_b)
mle = k_success / n_trials
samples = rng.beta(post_a, post_b, size=200_000)
print('posterior_mean_analytic', mean_p)
print('mle_k_over_n', mle)
print('mc_mean_from_beta_draws', float(samples.mean()))
```
