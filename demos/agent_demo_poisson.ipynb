# Companion demo: Poisson rate and a Wald CI

Draw IID **Poisson(λ)** counts; **MLE λ̂ = sample mean**. Build a Wald interval **λ̂ ± z·√(λ̂/n)** and print whether **λ** lands inside for this seeded draw.


```python
# WHY one fixed seed: reproducible stdout for `.apynb` drift checks.
import numpy as np
rng = np.random.default_rng(41)
```

```python
# WHY separate cell: isolation of sampling vs inference keeps partial runs readable.
lam_true = 3.8
n = 2500
counts = rng.poisson(lam_true, size=n)
```

```python
# WHY Wald SE sqrt(l_hat/n): large-n Poisson asymptotics (Variance ≈ λ for Poisson).
lam_hat = float(counts.mean())
se = float(np.sqrt(lam_hat / n))
z = 1.96
lo, hi = lam_hat - z * se, lam_hat + z * se
cover = float(lo <= lam_true <= hi)
print("lam_hat", lam_hat)
print("wald_ci", lo, hi)
print("covers_truth_round", cover)
```
