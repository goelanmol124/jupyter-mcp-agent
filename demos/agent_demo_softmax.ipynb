# Companion demo: softmax temperature vs entropy

Scale logits by **temperature** **T**. Low **T** sharpens (**argmax-ish**); high **T** flatten toward uniform (**higher Shannon entropy**). Fixed logits so runs are reproducible for companion checks.


```python
import numpy as np
```

```python
# WHY fixed logits: companion `.apynb` diffs stay stable versus random draws each run.
logits = np.array([2.3, -0.5, 0.9, -1.8, 0.2], dtype=float)
temperatures = np.array([0.25, 0.75, 1.0, 2.0, 8.0], dtype=float)
```

```python
def softmax(z: np.ndarray) -> np.ndarray:
    # WHY subtract max(z): avoids exp overflow and keeps softmax numerically stable.
    z = np.asarray(z, dtype=float)
    e = np.exp(z - np.max(z))
    return e / e.sum()


def categorical_entropy(p: np.ndarray) -> float:
    p = np.asarray(p, dtype=float)
    p = p[p > 1e-15]
    return float(-np.sum(p * np.log(p)))


# WHY loop (not broadcasting): readability in research cells beats micro-optimization here.
for T in temperatures:
    p = softmax(logits / float(T))
    print(f"T={T:g} entropy={categorical_entropy(p):.5f} max_prob={float(p.max()):.5f}")
```
