# Companion demo: rolling mean on a noisy random walk

Keeps **`agent_demo_buffer.apynb`** in sync; `run` validates when the sidecar exists.


```python
import numpy as np
rng = np.random.default_rng(3)
```

```python
x = rng.normal(0, 1, size=(200,)).cumsum()
window = 12
```

```python
roll = np.convolve(x, np.ones(window) / window, mode="valid")
print("last_smoothed", float(roll[-1]), "raw_last", float(x[-1]))
```
