# Changepoint toy (R&D pressure test)

Piecewise normal series: mean shifts at unknown index. Goal: recover break via max |cumulative mean residual| and cross-check intermediate shapes.

```
import numpy as np
np.random.seed(11)
n = 400
true_t = 210
seg1 = np.random.normal(0.0, 1.0, true_t)
seg2 = np.random.normal(1.8, 1.0, n - true_t)
y = np.concatenate([seg1, seg2])
print('n', len(y), 'true_t', true_t, 'mean0', seg1.mean(), 'mean1', seg2.mean())
```

```
mu = y.mean()
cums = np.cumsum(y - mu)
k = np.arange(1, len(y) + 1)
partial_means = cums / k
scores = np.empty(len(y) - 2)
for t in range(1, len(y) - 1):
    left = y[:t]
    right = y[t:]
    scores[t - 1] = abs(left.mean() - right.mean()) * np.sqrt(t * (len(y)-t) / len(y))
est = int(np.argmax(scores)) + 1
print('est_split', est, 'error', est - int(true_t))
print('score_peak', float(scores.max()))
```
