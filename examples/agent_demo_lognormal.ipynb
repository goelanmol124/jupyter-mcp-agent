# Companion demo: lognormal mean vs closed form

Sample **X ~ LogNormal(μ, σ²)** on the underlying normal parameterization NumPy uses, and compare the sample mean to **E[X] = exp(μ + σ²/2)**.


```python
import numpy as np
rng = np.random.default_rng(7)
mu, sigma = 0.3, 0.35
```

```python
n = 500_000
x = rng.lognormal(mean=mu, sigma=sigma, size=n)
```

```python
expected = float(np.exp(mu + 0.5 * sigma**2))
sample_mean = float(x.mean())
rel_err = abs(sample_mean - expected) / expected
print('sample_mean', sample_mean)
print('expected_E_X', expected)
print('relative_error', rel_err)
```
