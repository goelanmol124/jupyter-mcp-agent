# Companion demo: Exponential(rate) vs mean 1/λ

**Fact:** **E[X]** for **Exp(λ)** is **1/λ**. Compare the sample mean to the closed form (fixed seed).


```python
# Deterministic comparisons for `.apynb` drift checks rely on fixing the RNG seed here.
import numpy as np
rng = np.random.default_rng(53)
```

```python
# NumPy parametrization: `scale = 1/lambda` equals the mean.
lam = 0.73
n = 120_000
x = rng.exponential(scale=1.0 / lam, size=n)
```

```python
sample_mean = float(x.mean())
closed = float(1.0 / lam)
rel_err = abs(sample_mean - closed) / closed
print("sample_mean", sample_mean)
print("expectation_1_over_lambda", closed)
print("relative_error", rel_err)
```
