# ODE scratch: Euler vs exp(−t)

Tiny stability / error check for **`dy/dt = -y`**.


```python
import numpy as np
y = 1.0
dt = 0.08
n_steps = 60
```

```python
t = 0.0
for _ in range(n_steps):
    y = y + dt * (-y)
    t += dt
print("euler_t", t, "y_euler", y)
```

```python
y_true = float(np.exp(-t))
err = abs(y - y_true) / max(abs(y_true), 1e-12)
print("y_true", y_true, "relative_error", err)
```
