# Companion demo: two-state Markov stationary distribution

**Irreducible chain:** empirical occupation counts vs closed form \(\pi_0 = \frac{P_{10}}{P_{01}+P_{10}}\), \(\pi_1 = 1-\pi_0\).


```python
import numpy as np
rng = np.random.default_rng(17)
```

```python
P = np.array([[0.92, 0.08], [0.41, 0.59]])
assert np.allclose(P.sum(axis=1), 1.0)
pi0 = float(P[1, 0] / (P[0, 1] + P[1, 0]))
pi = np.array([pi0, 1.0 - pi0])
```

```python
steps = 350_000
state = int(rng.integers(0, 2))
counts = np.zeros(2, dtype=np.int64)
for _ in range(steps):
    counts[state] += 1
    u = rng.random()
    if u < P[state, 1]:
        state = 1
    else:
        state = 0
emp = counts.astype(float) / float(counts.sum())
print('pi_closed', pi)
print('empirical', emp)
print('abs_err', np.abs(pi - emp))
```
