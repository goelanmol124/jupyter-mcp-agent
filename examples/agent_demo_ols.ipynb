# Companion demo: recovering OLS coefficients under Gaussian noise

**Model:** linear **y = X @ beta + noise** with fixed truth; **`np.linalg.lstsq`** should recover **beta** when **n** is large enough.


```python
import numpy as np
rng = np.random.default_rng(23)
```

```python
n = 800
beta_true = np.array([1.25, -0.75])
x1 = rng.normal(size=n)
X = np.column_stack([np.ones(n), x1])
noise_sd = 0.35
y = X @ beta_true + noise_sd * rng.normal(size=n)
```

```python
coef, _, _, _ = np.linalg.lstsq(X, y, rcond=None)
rmse_coef = float(np.sqrt(np.mean((coef - beta_true) ** 2)))
print('beta_hat', coef)
print('beta_true', beta_true)
print('rmse_coef', rmse_coef)
```
